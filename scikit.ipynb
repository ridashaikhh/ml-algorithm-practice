Different ML Algo using SCIKIT LEARN 

Importing dependencies

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

```python
#taking the dataset from iris dataset
df = sns.load_dataset("iris")
```

```python
df.head()
```

```python
df.shape
```

Splitting the data into X and Y (feaures and labels)

```python
df.iloc[:,0:4]
```

```python
X = df.iloc[:,0:4].values
```

```python
print(X)
```

Encoding Species

```python
raw_y = df['species']
```

```python
labelencoder = LabelEncoder()
Y = labelencoder.fit_transform(raw_y)
```

```python
#Encoded Y(species)
print(Y)
```

Splitting the data into Train Test data

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2)
```

Training the model using various Algorithm.

* Logistic Regression
* KNeighbor Classifier
* Gaussian Naive Bayes
* Support vector classifier
    - kernel : Sigmoid, Linear, rbf, poly
* Decision Treee Classifier
* Random forest Classifier

```python
#Logistic Regression Model
from sklearn.linear_model import LogisticRegression
logmodel = LogisticRegression()
logmodel.fit(X_train, y_train)
```

```python
#Evaluating the model
y_logpred = logmodel.predict(X_test)
from sklearn.metrics import confusion_matrix
matrix = confusion_matrix(y_test, y_logpred)
print("The Confusion matrix : ")
print(matrix)
```

```python
26/30
```

```python
#KNeighborClassifier Algorithm

#Training the model
from sklearn.neighbors import KNeighborsClassifier
classifier_knn = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
classifier_knn.fit(X_train, y_train)
```

```python
#Evaluating the model
y_pred = classifier_knn.predict(X_test)
knmatrix = confusion_matrix(y_test, y_pred)
print("The Confusion matrix : ")
print(knmatrix)
```

```python
29/30
```

```python
#NAive Bayes Algorithm

#Training the model
from sklearn.naive_bayes import GaussianNB
classifier_nb = GaussianNB()
classifier_nb.fit(X_train,y_train)
```

```python
y_pred = classifier_nb.predict(X_test)
confusion_matrix(y_test, y_pred)
```

```python
#Support vector Classifier
#Kernel: Sigmoid

#Training the model
from sklearn.svm import SVC
classifier_svm_sigmoid = SVC(kernel='sigmoid')
classifier_svm_sigmoid.fit(X_train,y_train)
```

```python
#Evaluating
y_pred = classifier_svm_sigmoid.predict(X_test)
confusion_matrix(y_test, y_pred)
```
